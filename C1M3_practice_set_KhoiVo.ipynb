## Question 1: Filtering

Filter the mpg dataset to only include cars that are manufactured by toyota \(_i.e.,_ `manufacturer == "toyota"`\). use pipes, `%>%` and the `filter()` command. **Store the short dataframe in a new data frame named df\_toy.**


```r
# ✏️ Your code here

df_toy <- mpg %>%
  filter(manufacturer == "toyota")
```

```r
# glimpse your new dataframe
glimpse(df_toy)
```

## Question 2: Unit Conversion  

When working with data, you may need to convert the units. For example, in the `mpg` dataset, we can convert highway miles per gallon to kilometers per liter. Two necessary facts are:

- 1 mile = 1.61 kilometers
- 1 gallon = 3.79 liters

The unit transformation equation is

$$kpl = hwy /3.79 \times 1.61.$$

### Part A

Convert the variable `hwy` that measures highway miles per gallon to highway kilometers per liter. Use the `mutate()` function to create a new variable called `hwy_kpl`that measures highway kilometers per liter using the above unit transformation equation. Do so by starting with the original `mpg` dataset but storing the new dataset \(with the new variable `hwy_kpl`\) in a data frame called `df_metric`.


```r
# ✏️ Your code here
df_metric <- mpg %>%
  mutate(hwy_kpl = hwy / 3.79 * 1.61)
```

```r
# investigate your new df
summary(df_metric)
glimpse(df_metric)
```

### Part B

Find the average highway kilometers per liter for cars in your `df_metric` dataset. Store the value in a variable called `mean_kpl`.


```r
# ✏️ Your code here
mean_kpl <- mean(df_metric$hwy_kpl)
```

## Question 3: Summarizing Data

### Part A

Go back to using your `df_toy` dataset \(data frame that you filtered to toyota cars, only\). Group the `df_toy` data frame by the variable `class` and summarize the average high miles per gallon \(variable named `hwy`\) for each class. Store your summary dataframe in a new df called `df_sum`.


```r
# ✏️ Your code here

df_sum <- df_toy %>%
  group_by(class) %>%
  summarise(mean_hwy = mean(hwy))
```

### Part B

What class of Toyota car gets the best gas milage \(i.e., which gets the most miles per unit of gas used\)? Store the class of car that gets the best gas milage as a string/character variable in a variable named `best_class`.


```r
# ✏️ Your code here
best_class <- df_sum %>%
  filter(mean_hwy == max(mean_hwy)) %>%
  pull(class)
```

## Question Four

Let's finish off with some basic data visualization.

### Part A

Create a histogram of the `hwy` variable in the `mpg` dataset. Be sure to store your plot in an object called `plot_1`. Use ggplot. Note: it may warn you about `binwidth` but you can ignore this message.


```r
# ✏️ Your code here
plot_1 <- ggplot(data = mpg, aes(x = hwy)) +
  geom_histogram()
```

### Part B

Practice some data visualization best practice. Create a new plot called `plot_2` that is a histogram of the highway miles per gallon in the `mpg` dataset \(like Part A\), but do some of the following:

- write clear labels and a title
- make the theme as simple as possible
- make sure the text is legible
- consider changing colors \(is it legible for people who are color blind?\)


```r
# ✏️ Your code here
plot_2 <- ggplot(data = mpg, aes(x = hwy)) +
  geom_histogram(binwidth = 2, fill = "#0072B2", color = "white") +
  labs(
    title = "Distribution of Highway Fuel Economy",
    x = "Highway fuel economy (miles per gallon)",
    y = "Number of vehicles"
  ) +
  theme_classic(base_size = 14)
```
